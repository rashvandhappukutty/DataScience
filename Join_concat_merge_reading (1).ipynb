```python
import pandas as pd

# Titanic Dataset
titanic = pd.read_json("titanic.json")

print(titanic.head())

# Identify missing values
print(titanic.isnull())

# Count missing values
print(titanic.isnull().sum())

# Rows with missing Age
print(titanic[titanic["Age"].isnull()])

# Fill Age with mean
age_mean = titanic["Age"].mean()
titanic["Age"] = titanic["Age"].fillna(age_mean)

# Fill Embarked with mode
embarked_mode = titanic["Embarked"].mode()[0]
titanic["Embarked"] = titanic["Embarked"].fillna(embarked_mode)

# Remove rows with missing values
print(titanic.dropna())

# Remove columns with missing values
print(titanic.dropna(axis=1))

# iloc example
print(titanic.iloc[0:5, 1:3])

# GroupBy Operations
print("\nAverage Fare per Class")
print(titanic.groupby("Pclass")["Fare"].mean())

print("\nCount Survivors by Gender")
print(titanic.groupby("Sex")["Survived"].sum())

print("\nMultiple Aggregations")
print(
    titanic.groupby("Pclass").agg({
        "Age": ["mean", "min", "max"],
        "Fare": ["mean", "sum"]
    })
)

# Customer Dataset
customers = pd.read_csv("customers-100.csv", index_col="Customer Id")

print(customers.head())

# Access specific columns for particular index
print(
    customers.loc[
        "053d585Ab6b3159",
        ["First Name", "Company"]
    ]
)

# Range of records
print(
    customers.loc[
        "cb8E23e48d22Eae":"2354a0E336A91A1",
        ["First Name", "Company"]
    ]
)
```

## INNER MERGE 

```python
import pandas as pd

students = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="students")

marks = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")

contact = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")

#1. INNER MERGE
inner_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

print(inner_merge.head())
print(inner_merge.shape)
```

## Combine Students , Marks , Contact

```python
import pandas as pd

students = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="students")

marks = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")

contact = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")

# Combine Students, Marks, and Contact

combined_data = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

combined_data = pd.merge(
    combined_data,
    contact,
    on="Student_ID",
    how="left"
)

print(combined_data.head())
print(combined_data.shape)
```

## Concat Column Wise

```python
# 7. CONCAT COLUMN-WISE

student_basic = students[['Student_ID', 'Name']]
student_marks = marks[['Python']]

combined_cols = pd.concat(
    [student_basic, student_marks],
    axis=1
)

print(combined_cols.head())
```

## Concat Row Wise


```python
part1 = students.iloc[:25]
part2 = students.iloc[25:]

print(part1.shape)
print(part2.shape)

# Concat Row-Wise

combined_rows = pd.concat(
    [part1, part2],
    axis=0
)

print(combined_rows.head())
print(combined_rows.shape)
```

## Join () two datasets by using index

```python
# JOIN()
# join() works using indexes

students_index = students.set_index("Student_ID")
contact_index = contact.set_index("Student_ID")

joined_df = students_index.join(contact_index)

print(joined_df.tail(10))
```
