```python
import pandas as pd
from scipy.stats import zscore

# Load dataset
df = pd.read_csv("Messy_Employee_dataset.csv")

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# -----------------------------------
# 1. Descriptive Statistics
# -----------------------------------
descriptive_stats = numeric_df.describe()

print("Descriptive Statistics:")
print(descriptive_stats)

# -----------------------------------
# 2. Identify Skewed Distributions
# -----------------------------------
skewness = numeric_df.skew()

print("\nSkewness Values:")
print(skewness)

for col, value in skewness.items():
    if value > 0.5:
        print(f"{col}: Right-Skewed")
    elif value < -0.5:
        print(f"{col}: Left-Skewed")
    else:
        print(f"{col}: Approximately Symmetric")

# -----------------------------------
# 3. Z-Score Normalization
# -----------------------------------
zscore_df = numeric_df.apply(
    lambda x: zscore(x, nan_policy='omit')
)

print("\nZ-Score Normalized Data:")
print(zscore_df.head())

# -----------------------------------
# 4. Pearson Correlation
# -----------------------------------
pearson_corr = numeric_df.corr(method='pearson')

print("\nPearson Correlation Matrix:")
print(pearson_corr)

# -----------------------------------
# 5. Spearman Correlation
# -----------------------------------
spearman_corr = numeric_df.corr(method='spearman')

print("\nSpearman Correlation Matrix:")
print(spearman_corr)

# -----------------------------------
# 6. Save Results
# -----------------------------------
with pd.ExcelWriter("Statistical_Analysis.xlsx") as writer:
    descriptive_stats.to_excel(writer, sheet_name="Descriptive_Stats")
    skewness.to_frame("Skewness").to_excel(writer, sheet_name="Skewness")
    zscore_df.to_excel(writer, sheet_name="ZScore_Normalized", index=False)
    pearson_corr.to_excel(writer, sheet_name="Pearson_Correlation")
    spearman_corr.to_excel(writer, sheet_name="Spearman_Correlation")

print("\nAnalysis saved successfully!")
```

```python
'''Method 1: Visual Detection Using Box Plot

A Box Plot quickly shows:

Minimum Value (excluding outliers)
First Quartile (Q1) – 25th percentile
Median (Q2) – 50th percentile
Third Quartile (Q3) – 75th percentile
Maximum Value (excluding outliers)
Outliers – data points lying outside the whiskers

A box plot helps visually identify the spread of data and potential outliers.

Method 2: IQR (Interquartile Range) Method

Formula: IQR = Q3 - Q1

Steps to Detect Outliers:

Calculate Q1 and Q3
Find IQR = Q3 − Q1
Compute:
Lower Bound = Q1 − 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR

Any value:

Less than the Lower Bound, or
Greater than the Upper Bound

is considered an outlier.

Example:

If Q1 = 20 and Q3 = 40,

IQR = 40 − 20 = 20
Lower Bound = 20 − (1.5 × 20) = -10
Upper Bound = 40 + (1.5 × 20) = 70

Any value below -10 or above 70 is an outlier.''
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import zscore

#Load Titanic dataset
df = sns.load_dataset('titanic')
# Box Plot
sns.boxplot(x = df['fare'])
plt.show()

# IQR method
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers_iqr = df[
    (df['fare'] < lower) |
    (df['fare'] > upper)]
print("IQR Outliers:",len(outliers_iqr))
# Z Scores
df['z_score'] = zscore(df['fare'])

outliers_z = df[
    abs(df['z_score']) > 3
]

print("Z-Score Outliers:", len(outliers_z))

```

```python
# Outlier Detection Using IQR Method
df = sns.load_dataset('tips')

# Select column
data = df['total_bill']

# Calculate Q1 and Q3
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(data < lower_bound) | (data > upper_bound)]
# Display results
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
print("Number of Outliers =", len(outliers))
```
