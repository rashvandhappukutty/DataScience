## BOX PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt

salary = [25000, 20000, 30000, 32000, 35000,
          37000, 40000, 42000, 45000, 60000]

plt.boxplot(salary)

plt.title("Employee Salaries")
plt.xlabel("Employee")
plt.ylabel("Salary")

plt.show()
```

```python
"""Box plot
A box plot is used to visualize the distribution of numerical data and identify outliers.
components of a box plot
Min--- Q1---[Median]---Q3-----Max
    |                      |
Lower whisker        Upper Whisker
Minimum:Smallest value(excluding outliers)
Q1(25%): First quartile
Median (50%): Middle value
Q3(75%): Third quartile
Maximum: Largest value(excluding outliers)
Outliers: Values far from the rest of the data, shown as individual points
```

## Heatmap

```python
# Correlation Heatmap with Seaborn

import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

# Correlation matrix
corr = tips.corr(numeric_only=True)

# Heatmap
sns.heatmap(corr, annot=True)

# Title
plt.title("Correlation Heatmap")

# Display
plt.show()
```

```python
"""Heatmap what is a heatmap?
 A heatmap uses colors to represent values in a table or matrix.
 In data analysis, heatmaps are commonly used to visualize correlations between numerical columns.

 correlation values
 Value meaning
 +1  -> Strong Positive Relationship
 0   -> No Relationship
 -1  -> Strong Negative Relationship
 """
```

## Histogram

```python
# Histogram with Seaborn

import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

# Histogram
sns.histplot(tips["total_bill"], bins=10)

# Title
plt.title("Histogram of Total Bill")

# Labels
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

# Display
plt.show()
```

## KDE

```python
sns.kdeplot(
    data=tips,
    x="total_bill"
)
plt.title("KDE Plot")
plt.show()
```

```python
'''KDE Plot (Kernal Density Estimation) what is KDE plot?
A KDE plot is a smooth version of a histogram
Instead of bars,it shows a smooth curve representing data distribution
When to use KDE? Understanding data distribution identifying peaks comparing distribution statistical analysis'''
```

## Histogram with KDE

```python
sns.histplot(
    data=tips,
    x= "total_bill",
    kde=True
    )
    plt.title("Histogram with KDE")
```

## SubPlot

### Example 1: Two Charts

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))

# First Plot
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("Line Chart")

# Second Plot
plt.subplot(1, 2, 2)
plt.bar(['A', 'B', 'C'], [5, 8, 6])
plt.title("Bar Chart")

plt.show()
```

### Example 2: Four Charts


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

# Line Chart
plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("Line")

# Bar Chart
plt.subplot(2, 2, 2)
plt.bar(['A', 'B', 'C'], [5, 8, 6])
plt.title("Bar")

# Scatter Chart
plt.subplot(2, 2, 3)
plt.scatter([1, 2, 3], [4, 8, 6])
plt.title("Scatter")

# Pie Chart
plt.subplot(2, 2, 4)
plt.pie([30, 40, 30], labels=['A', 'B', 'C'])
plt.title("Pie")

plt.tight_layout()
plt.savefig(
    "chart.png",
    dpi = 300
)
plt.show()
```
