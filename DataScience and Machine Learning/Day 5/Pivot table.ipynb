```python
"""A Pivot table is used to summarize and analyse data by grouping rows and applying aggregate functions such as sum, mean, count, max, min etc."""

# Step 1: Load dataset
import pandas as pd

df = pd.read_csv("people-100.csv")
print(df.head())

# Step 2: Create Age column
df['Date of birth'] = pd.to_datetime(df['Date of birth'])

today = pd.Timestamp.today()

df['Age'] = ((today - df['Date of birth']).dt.days // 365)

print(df[['First Name', 'Sex', 'Age']].head())

# Step 3: Create Pivot Table
pivot_table = pd.pivot_table(
    df,
    values='Age',
    index='Sex',
    aggfunc=['mean', 'max', 'min', 'count']
)

print(pivot_table)
```

```python
# 4. Pivot Table with aggfunc = 'count'

pivot4 = pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc='count'
)

print(pivot4)
```

```python
# 5. Pivot Table with aggfunc = 'max'

pivot5 = pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc='max'
)

print(pivot5)
```

```python
# 6. Pivot Table with aggfunc = 'min'

pivot6 = pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc='min'
)

print(pivot6)
```

```python
# 7. Multiple Aggregate Functions

pivot7 = pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc=['mean', 'sum', 'count', 'max', 'min']
)

print(pivot7)
```

```python
# 8. Pivot Table with Margins (Grand Total)

pivot8 = pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc='mean',
    margins=True
)

print(pivot8)
```

```python
#  9. Overall Average Age using Pivot Table

overall_avg = pd.pivot_table(
    df,
    values="Age",
    aggfunc="mean"
)

print(overall_avg)
```

```python
# 10. Pivot Table: Average Age by Job Title and Gender

pivot10 = pd.pivot_table(
    df,
    values="Age",
    index=["Job Title", "Sex"],
    aggfunc="mean"
)

print(pivot10)
```

```python
# 10. Pivot Table: Average Age by Job Title and Gender

pivot11 = pd.pivot_table(
    df,
    values='User Id',
    columns='Sex',
    aggfunc='count'
)

print(pivot11)
```
