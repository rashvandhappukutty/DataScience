## Reading the file 

```python
import pandas as pd
df = pd.read_excel("Data_Cleaning_50Plus_Records.xlsx")
print(df.head())
```

## Rename the Coloums 

```python
import pandas as pd

# Rename columns
df.rename(columns={
    'Student ID': 'Student_ID',
    'Student Name ': 'Student_Name',
    ' Department': 'Department'
}, inplace=True)

print(df.columns)
```

## Checking the Data Types

```python
print(df.dtypes)
```

## Converting Student ID into String

```python
# Convert Student_ID to String

df['Student_ID'] = df['Student_ID'].astype(str)

# Convert Age to Integer
# First handle missing values

df['Age'] = df['Age'].fillna(0)

df['Age'] = df['Age'].astype(int)

# Check Data Types

print(df.dtypes)
```

## Detect Duplicate Records

```python
# Detect Duplicate Records

duplicates = df.duplicated()

print(duplicates)
```

# DAY 5 Data Cleaning

## Count Duplicates

```python
print(df.duplicated().sum())
print(df.columns)
```

## Dropping Duplicate Records

```python
# Before
print(df.shape)

# Remove Duplicates
df.drop_duplicates(inplace=True)

# After
print(df.shape)
```

## String Operations 

```python
# String methods work only on text columns
# Remove Extra Spaces
df['Student_Name'] = df['Student_Name'].str.strip()

# Convert to Uppercase
df['Student_Name'] = df['Student_Name'].str.upper()

# Convert to Title Case
df['Student_Name'] = df['Student_Name'].str.title()

# Display first 5 records
print(df['Student_Name'].head())
```

```python
# Replace Text 

# Replace department

df['Department'] = df['Department'].replace({
    'CSE': 'Computer Science',
    'ECE': 'Electronics and Communication Engineering',
    'EEE': 'Electrical and Electronics Engineering',
    'IT': 'Information Technology',
    'MECH': 'Mechanical Engineering'
})

print(df['Department'].head())
```
